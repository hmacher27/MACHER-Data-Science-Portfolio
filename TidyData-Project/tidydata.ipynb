```python
# Import the libraries I need for cleaning and plotting
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("fed_rd_year&gdp.csv")
df
```

```python
df.info()
df.describe()
df.isna().sum()
```

## Inspect the Data

Before changing anything, I ran `df.info()` and `df.isna().sum()` to get a feel for how the dataset is set up.

**Why this data is untidy:** The column names put two variables together, the year and the GDP. That breaks the tidy data rule that each variable gets its own column. To fix it, I'll use `melt()` to go from wide to long, then `str.split()` to pull the year and GDP apart.

## Reshaping the Data with `melt()`

Right now the data is in wide format, where every year is its own column. 

I used `melt()` to switch it to long format. I kept `department` as the identifier, put the old column names into a new `year_gdp` column, and put the spending values into a new `rd_budget` column. After this, each row is one department in one year.

The `year_gdp` column still has two variables stuck together (year and GDP), so I'll split it next.

```python
# Melt from wide to long 
df_long = df.melt(
    id_vars = "department",
    var_name = "year_gdp",
    value_name = "rd_budget"
)
df_long.head()
```

## Tidy the Data

After melting, `year_gdp` still holds **two variables** (year and GDP) in one column. That breaks the first tidy data principle: *each variable forms its own column*. I'll split it into separate `year` and `gdp` columns.

```python
# Split strings into two columns at "_gdp"
df_long[["year", "gdp"]] = df_long["year_gdp"].str.split("_gdp", expand=True)

# str.split gives back strings, so convert to real numeric types
df_long["year"] = df_long["year"].astype(int)
df_long["gdp"] = df_long["gdp"].astype(float)

# Drop the old combined column since we don't need it anymore
df_long = df_long.drop(columns="year_gdp")

df_long.head()
```

## Handle Missing Values

Some departments (like DHS) have NaN values for the early years. These aren't zeros, the department just didn't exist yet (DHS was created in 2002-2003). If I kept them, the data would make it look like those departments existed and spent nothing, so I'm dropping those rows.

```python
# See how many missing values there are and which departments they belong to
print(df_long["rd_budget"].isna().sum())
print(df_long[df_long["rd_budget"].isna()]["department"].unique())

# Drop rows where the department didn't exist yet
df_long = df_long.dropna(subset=["rd_budget"]).reset_index(drop=True)
```

 ## Separate Observational Units into Their Own Tables

The third tidy data principle says *each type of observational unit forms its own table*. GDP is a property of the **year**, not of the department, so right now it's getting repeated on every department row. To fix that, I'm splitting the data into two tables:

- `gdp_by_year`: one row per year
- `rd_by_dept`: one row per department per year

```python
# Table 1: GDP, one row per year
gdp_by_year = df_long[["year", "gdp"]].drop_duplicates().reset_index(drop=True)

# Table 2: R&D budget, one row per department per year
rd_by_dept = df_long[["department", "year", "rd_budget"]].copy()

print(gdp_by_year.shape, rd_by_dept.shape)
gdp_by_year.head()
```

## Validate the Tidy Data

Now I'm running a few final checks to make sure there aren't any data quality problems left.

```python
rd_by_dept.info()
print("Missing values:\n", rd_by_dept.isna().sum())
print("Duplicate rows:", rd_by_dept.duplicated().sum())
rd_by_dept.head()
```

```python
merged = rd_by_dept.merge(gdp_by_year, on="year")
merged["rd_pct_gdp"] = merged["rd_budget"] / merged["gdp"] * 100
```

## Combine Tables for Analysis

Since each observational unit has its own table now, I can merge them back together on `year` for the analysis. This lets me create a new variable so spending is adjusted for the size of the economy, which makes comparing different years fairer.

```python
# Merge the two tables on year
merged = rd_by_dept.merge(gdp_by_year, on="year")

# New variable
merged["rd_pct_gdp"] = merged["rd_budget"] / merged["gdp"] * 100

# Convert to billions
merged["rd_budget_billions"] = merged["rd_budget"] / 1e9

merged.head()
```

## Visualization 1 - Total Federal R&D Over Time

I used a line plot because it's the best way to show how something changes over time.

```python
# Add up R&D budget across all departments for each year
total_by_year = merged.groupby("year")["rd_budget_billions"].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.lineplot(data=total_by_year, x="year", y="rd_budget_billions", marker="o")
plt.title("Total Federal R&D Budget Over Time (1976-2017)")
plt.xlabel("Year")
plt.ylabel("R&D Budget (Billions of Dollars)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Visualization 2 - R&D Budget by Department

Next I plotted each department as its own line to see which agencies are driving the overall trend. This was easy to do because the data is tidy. 

```python
# One line per department
plt.figure(figsize=(12, 7))
sns.lineplot(data=merged, x="year", y="rd_budget_billions", hue="department")
plt.title("Federal R&D Budget by Department (1976-2017)")
plt.xlabel("Year")
plt.ylabel("R&D Budget (Billions of Dollars)")
# Put the legend outside the plot so it doesn't cover the lines
plt.legend(title="Department", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Visualization 3 - R&D as a Percentage of GDP

Raw dollar amounts don't account for the economy growing over time. By using the `rd_pct_gdp` variable I made earlier, I can see whether R&D spending is actually keeping up with the economy.

```python
# Total R&D as a percent of GDP for each year
pct_by_year = merged.groupby("year")["rd_pct_gdp"].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.lineplot(data=pct_by_year, x="year", y="rd_pct_gdp", marker="o", color="darkgreen")
plt.title("Total Federal R&D as a Percentage of GDP")
plt.xlabel("Year")
plt.ylabel("R&D (% of GDP)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Pivot Table / Aggregation

I made a pivot table that summarizes the tidy data by department. It shows the average, peak, and total R&D spending over the whole time period.

```python
# Mean, max, and total R&D spending for each department
pivot = merged.pivot_table(
    index="department",
    values="rd_budget_billions",
    aggfunc=["mean", "max", "sum"]
).round(2)

# Rename the columns 
pivot.columns = ["mean_billions", "max_billions", "total_billions"]

# Sort so biggest spenders are at the top
pivot.sort_values("total_billions", ascending=False)
```

## Conclusion

**Key findings:**
- **Which departments spend the most:** The Department of Defense (DOD) is the biggest spender. Its total across 1976-2017 was about \$2,717 billion, with an average of around \$64.7 billion a year and a peak of about \$94.3 billion. HHS (about \$936 billion) and NIH (about \$887 billion) come next, followed by NASA (about \$510 billion) and DOE (about \$499 billion). On the department plot, DOD sits way above everything else, and a lot of the smaller agencies (EPA, DOT, VA, DHS, etc.) are squished together near the bottom, each under \$2 billion a year.
- **How total spending changed:** Overall, total federal R&D went up a lot, from about \$86 billion in 1976 to a peak of about \$205 billion around 2010. The two biggest jumps were in the mid-1980s and from about 2001 to 2004, when it went from around \$142 billion to almost \$200 billion. After 2010, spending dropped to around \$172 billion by 2013, bounced back up in 2016, then fell to about \$164 billion in 2017. The department plot shows DOD is behind most of these swings. It had a peak in the late 1980s, a dip in the 1990s, a huge climb in the 2000s (to about \$94 billion), and then a sharp drop after 2010. HHS and NIH grew steadily from the late 1990s through the early 2000s and then stayed pretty flat.
- **R&D vs. the economy:** Even though the dollar amounts mostly went up, R&D as a share of GDP went down almost the whole time. It started at about 4.8% in 1976 and fell to under 1% (about 0.85%) by 2017. The only real bump was around 2001-2003, when it rose to about 1.7% before falling again. So the economy grew much faster than federal R&D spending did, and the rise in dollars doesn't mean R&D kept up with the economy.

The totals add up every department, and NIH is part of HHS, so some spending is probably counted twice. That could make the total and % of GDP lines a bit high, but the overall trends should still hold. Also, the 2017 values look unusually low for a few departments (like DOD and VA), so some of that last-year drop might be from incomplete data and not a real cut.

Getting the data into a tidy format made the rest of the project a lot easier. Since year and department each had their own column, I could use `groupby()`, `hue="department"`, and `pivot_table()` without having to rewrite anything. If I had stayed in the original wide format with 42 year columns, every one of these steps would have been much more annoying. Splitting GDP into its own table also kept it from being repeated on every department row, which is what the third tidy data principle is about.
