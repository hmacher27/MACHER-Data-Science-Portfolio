## Step 1: Inspect the Data

Before cleaning, I used `df.info()` and `df.isna().sum()` to understand the structure of the dataset.

**Findings:**
- **Shape:** The dataset has 14 rows (one per federal department) and 43 columns (one `department` column plus 42 year columns covering 1976-2017).
- **Missing values:** Each year from 1976 to 2001 has exactly one missing value, while 2002-2017 have none. This is likely the Department of Homeland Security (DHS), which was created in 2002. These blanks mean the department did not exist yet, not that spending was zero, so I will drop them rather than fill them in.
- **Data types:** Columns with missing values are stored as `float64` because pandas cannot store NaN in integer columns. Complete columns are `int64`. This explains the type change at 2001/2002.

**Why the data is untidy:** The column names combine two variables, the year and the GDP (for example, `1976_gdp1790000000000.0`). This violates the tidy data principle that each variable should have its own column. In the next steps, I will use `melt()` to reshape the data from wide to long, then `str.split()` to separate the year and GDP into their own columns.

## Step 2: Reshape the Data with `melt()`

The dataset is in "wide" format: each year is its own column, which means the year (a variable) is stored in the column headers instead of in a column of its own. This violates the tidy data principle that each variable should form a column.

I used `melt()` to convert the data from wide to long format. The `department` column is kept as an identifier, the old column names are stored in a new `year_gdp` column, and the spending values are stored in a new `rd_budget` column. After melting, each row is one department in one year.

The `year_gdp` column still contains two variables combined (the year and the GDP), so it will need to be split in the next step.

```python
df_long = df.melt(
    id_vars = "department",
    var_name = "year_gdp",
    value_name = "rd_budget"
)
df_long.head()
```
